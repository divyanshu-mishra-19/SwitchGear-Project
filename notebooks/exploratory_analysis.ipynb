# CircuitGuard AI: Exploratory Data Analysis & Model Evaluation
**AI-Based Predictive Maintenance of High-Voltage Circuit Breakers Using Machine Learning**

This notebook documents the exploratory data analysis, feature distributions, missingness patterns, and model evaluation metrics for the HV Circuit Breaker dataset.

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

raw_df = pd.read_excel("../dataset/raw/HV Circuit Breaker Maintenance Data.xlsx")
print("Dataset Shape:", raw_df.shape)
raw_df.head()
```

## 1. Target Class Distribution

```python
print(raw_df["Maintenance_required"].value_counts())
print(raw_df["Maintenance_required"].value_counts(normalize=True))
```

## 2. Structural Missing Values by Product Variant

```python
nulls_by_var = raw_df.groupby("Product_variant").apply(lambda g: g.isna().sum(), include_groups=False)
cols_with_nulls = nulls_by_var.columns[nulls_by_var.sum() > 0]
nulls_by_var[cols_with_nulls]
```

## 3. Untouched Test Set Evaluation Metrics

```python
with open("../reports/model_metrics.json", "r") as f:
    metrics = json.load(f)
metrics_df = pd.DataFrame(metrics).T[["accuracy", "precision", "recall", "f1_score", "roc_auc", "pr_auc"]]
metrics_df.sort_values(by="f1_score", ascending=False)
```
